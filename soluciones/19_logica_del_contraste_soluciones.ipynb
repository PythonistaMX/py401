## 2. La lógica del contraste.

Una prueba de hipótesis es una pieza de una investigación más amplia. Antes de calcular un p-value hay que saber cómo se obtuvieron los datos, porque eso decide qué se puede concluir.

### 2.1 Del método científico a la prueba de hipótesis.

**Concepto: el ciclo de una investigación.**
1. **Pregunta:** qué se quiere saber; por ejemplo, ¿los machos Adelie pesan más que las hembras?
2. **Hipótesis:** la pregunta se traduce en afirmaciones sobre la población que los datos puedan contradecir.
3. **Diseño del estudio:** cómo se obtendrán los datos (con un **experimento** o con un **estudio observacional**) y cómo se elegirán los individuos.
4. **Datos y análisis:** se describen los datos y se aplica la prueba.
5. **Conclusión y alcance:** qué dicen los resultados, sobre quién y si permiten hablar de causa.

Las pruebas de este módulo son el paso 4; los pasos 3 y 5 deciden cómo se redacta la conclusión.

**Concepto: experimento contra estudio observacional.**
* En un **experimento**, quien investiga **asigna** el tratamiento a cada individuo. Si lo asigna **al azar** (con un sorteo), los grupos quedan, en promedio, parecidos en todo lo demás, incluso en lo que no se midió. Así, una diferencia que el azar difícilmente produciría se puede atribuir al tratamiento: **la asignación al azar permite hablar de causa**.
* En un **estudio observacional** solo se registra lo que ya ocurrió: nadie decide quién es macho, quién viaja en tercera clase o quién fuma. Los grupos pueden diferir en otras variables (los **confusores**), así que una diferencia muestra una **asociación**, no una causa.
* Otra pregunta es **a quién** se aplica la conclusión. Si los individuos se eligieron **al azar** de una población, el resultado se puede extender a esa población: **el muestreo al azar permite generalizar**. Si no, el resultado describe a los individuos estudiados y, con cautela, a otros que se les parezcan.

**Ejemplo:** para saber si un recordatorio por mensaje aumenta la asistencia a las citas de una clínica, se sortea entre los pacientes quién lo recibe. Si los que lo recibieron asisten más, se puede decir que el recordatorio **causó** parte de la diferencia. Si solo se compara a quienes activaron los recordatorios por su cuenta con quienes no lo hicieron, la diferencia podría deberse a que los primeros ya eran más organizados.

**El alcance de la inferencia.**

| | **Con asignación al azar** (experimento) | **Sin asignación al azar** (observacional) |
|---|---|---|
| **Con muestreo al azar** | Causa, y se generaliza a la población | Asociación, y se generaliza a la población |
| **Sin muestreo al azar** | Causa, para los individuos estudiados | Asociación, para los individuos estudiados |

Los tres conjuntos de este módulo (`penguins`, `titanic` y `tips`) son **observacionales**: nadie asignó la especie, el sexo, la clase ni el hábito de fumar. Por eso sus conclusiones se redactan como asociaciones («se asocia con» y no «causa»). Para generalizar hay que preguntarse, además, cómo se eligieron los individuos: los pasajeros del Titanic, por ejemplo, no son una muestra al azar de ningún otro viaje.

> 🤔 **Explícalo con tus palabras.** ¿Por qué el sorteo del tratamiento permite hablar de causa, aunque haya variables que nadie midió? ¿Por qué un muestreo al azar, por sí solo, no lo permite?

**✅ Una explicación posible:** el sorteo no depende de ninguna característica de los individuos, así que, en promedio, reparte por igual entre los grupos todo lo demás, se haya medido o no: lo único que distingue de forma sistemática a los grupos es el tratamiento. El muestreo al azar hace que la muestra se parezca a la población, pero si nadie asigna el tratamiento, los grupos pueden seguir difiriendo en confusores.

### 2.2 Las hipótesis: una H0 y una H1 por prueba.

**Concepto: qué son H0 y H1.**
* Las dos son afirmaciones sobre la **población** (una media, una proporción, una diferencia), no sobre la muestra: de la muestra ya se sabe todo.
* **H0 (hipótesis nula)** es la afirmación de «no hay efecto» y fija un valor exacto; por ejemplo, $\mu_{\text{machos}} - \mu_{\text{hembras}} = 0$. Ese valor exacto es el que permite calcular qué datos se esperarían si H0 fuera cierta.
* **H1 (hipótesis alternativa)** es lo que se concluye si los datos contradicen a H0, y cubre lo que H0 no cubre: la diferencia no es 0 (bilateral), o es mayor que 0 (unilateral, solo si la dirección se decidió **antes** de ver los datos).
* **Cada prueba tiene una sola H0 y una sola H1**, escritas antes de ver los resultados. Si hay varias preguntas (varias variables, varios subgrupos, varios pares de grupos), hay **varias pruebas**, cada una con su H0 y su H1, y cada prueba agrega una oportunidad de falsa alarma. Ese es el problema de las **comparaciones múltiples**, que se simula más abajo.
* Con tres o más grupos (ANOVA, en `21_proporciones_y_mas_de_dos_grupos.ipynb`) también hay una sola H0 y una sola H1: H0 dice que todas las medias son iguales y **H1 dice que al menos una media difiere**, no que todas sean distintas entre sí. Para saber cuáles difieren hacen falta otras pruebas, por pares y con corrección.

**Ejemplo:** una pregunta, una prueba, una H0 y una H1.

| Pregunta | H0 | H1 |
|---|---|---|
| ¿Difiere la masa media de machos y hembras Adelie? | La diferencia de medias en la población es 0 | La diferencia no es 0 |
| ¿Difiere la supervivencia de mujeres y hombres en el Titanic? | Las dos proporciones de supervivencia son iguales | Las proporciones son distintas |
| ¿Se asocian la clase y la supervivencia? | Clase y supervivencia son independientes | No son independientes |
| ¿Difiere la masa media entre las tres especies? | Las tres medias son iguales | Al menos una media difiere |

### 2.3 El p-value y los dos errores.

| Concepto | Significado |
|---|---|
| H0 (hipótesis nula) | "No hay efecto": diferencia 0, independencia, medias iguales |
| H1 (alternativa) | Hay diferencia (bilateral) o va en una dirección (unilateral, decidida **antes** de ver los datos) |
| p-value | Probabilidad, **si H0 fuera cierta**, de obtener un resultado al menos tan extremo como el observado |
| α (nivel de significancia) | Umbral fijado de antemano (habitualmente 0.05) |
| Error tipo I | Rechazar H0 siendo cierta (probabilidad α) |
| Error tipo II | No rechazar H0 siendo falsa (probabilidad β) |
| Potencia | 1 − β: probabilidad de detectar un efecto que sí existe |

**Advertencia:** el p-value **no** es la probabilidad de que H0 sea cierta, no mide el tamaño del efecto y *p* > 0.05 **no** demuestra que no haya diferencia.

**Concepto: los dos errores, con una alarma de humo.** Piensa en una alarma de humo; su hipótesis nula (H0) es «no hay incendio».
* El **error tipo I** es que suene sin incendio: una falsa alarma. Su probabilidad es α, que se fija de antemano.
* El **error tipo II** es que **no** suene cuando sí hay incendio. La **potencia** es la probabilidad de que sí suene cuando sí hay incendio (1 − β).

Bajar α para tener menos falsas alarmas hace más probable que la alarma se calle ante un incendio pequeño: los dos errores se compensan, salvo que se tengan más datos (un detector mejor): con más datos, β baja sin que α suba.

| | **H0 es cierta** (no hay efecto) | **H0 es falsa** (sí hay efecto) |
|---|---|---|
| **No se rechaza H0** | Decisión correcta | Error tipo II (probabilidad β) |
| **Se rechaza H0** | Error tipo I (probabilidad α) | Decisión correcta (potencia = 1 − β) |

**¿Por qué 0.05?** Es una convención, no una ley: significa aceptar que, de cada 20 veces que no hay nada, se declare un hallazgo una. En contextos de más riesgo se usa un umbral menor.

**Ejemplo de «*p* > 0.05 no demuestra que no haya diferencia»:** los Chinstrap y los Adelie difieren en masa media por solo 32 g (3,733 contra 3,701 g) y la prueba da *p* = 0.59. Pero el intervalo de confianza de la diferencia (calculado con el método de Welch) va de −86 a +150 g: los datos no descartan diferencias de hasta unos 150 g.

**Concepto: cómo se construye una prueba.** Toda prueba de hipótesis sigue los mismos cuatro pasos:

1. Plantear H0 y H1 **antes** de ver los resultados.
2. Elegir un **estadístico de prueba** que mida qué tan lejos están los datos de lo que predice H0. Por ejemplo, para dos medias, el estadístico *t* es la diferencia observada entre las medias **medida en errores estándar**: una *t* de 3 quiere decir que la diferencia es tres veces más grande que lo que el azar suele mover a una diferencia así.
3. Averiguar qué valores tomaría ese estadístico **si H0 fuera cierta** (con una fórmula, como la *t* de Student, o por simulación, como la permutación).
4. Calcular el p-value: la probabilidad, si H0 fuera cierta, de obtener un estadístico al menos tan alejado de cero como el observado. Si es menor que α (habitualmente 0.05), se rechaza H0.

**Ejemplo cotidiano:** un amigo dice que su moneda es normal (H0) y sale águila 10 veces seguidas. Si la moneda fuera normal, ver 10 águilas seguidas o 10 soles seguidos (igual de sospechoso) pasaría unas 2 veces en 1,000: ese es el p-value. No prueba que haga trampa, pero hace difícil creerle.

Si H0 es cierta y el estadístico de la prueba es continuo (como la *t* con datos de masa), el p-value puede salir con igual probabilidad en cualquier punto entre 0 y 1: hay 5% de probabilidad de que salga menor que 0.05 por puro azar. Eso es exactamente el error tipo I, y es lo que explota el *p-hacking*.

> 💡 **Rechazar no es lo mismo que aceptar.** Cuando *p* es grande no se «acepta» H0: se dice que los datos no muestran evidencia en su contra. Es como un juicio: la falta de pruebas no demuestra inocencia.

**Ejemplo: el p-value a la vista, por permutación.** Para ver qué significa «si H0 fuera cierta», se **revuelven al azar las etiquetas** macho y hembra de los 146 Adelie con sexo registrado y se calcula la diferencia de masa media en cada revuelta. Si H0 fuera cierta (es decir, si el sexo no se asociara con la masa de los Adelie), esas diferencias mostrarían lo que el azar puede producir por sí solo. Se hacen 10,000 revueltas y se compara con la diferencia observada.

* La siguiente celda revolverá 10,000 veces las etiquetas de sexo de los Adelie, dibujará las diferencias obtenidas junto a la diferencia observada (línea negra) y mostrará cuatro números: la diferencia observada, la desviación estándar de las revueltas, los límites entre los que quedó el 95% central de las revueltas y la mayor diferencia (en valor absoluto) que produjo el azar.

```python
adelie = pinguinos[pinguinos["species"] == "Adelie"].dropna(subset=["body_mass_g", "sex"])
masa_adelie = adelie["body_mass_g"].to_numpy()
es_macho = (adelie["sex"] == "Male").to_numpy()
diferencia_observada = masa_adelie[es_macho].mean() - masa_adelie[~es_macho].mean()

generador_permutacion = np.random.default_rng(401)
diferencias_revueltas = np.empty(10_000)
for i in range(10_000):
    revuelto = generador_permutacion.permutation(es_macho)
    diferencias_revueltas[i] = masa_adelie[revuelto].mean() - masa_adelie[~revuelto].mean()

ax = sns.histplot(diferencias_revueltas, bins=50)
ax.axvline(diferencia_observada, color="black")
ax.set(xlabel="diferencia de masa media (g) con las etiquetas revueltas",
       title="Lo que el azar produce si H0 fuera cierta (línea negra: diferencia observada)")
plt.show()
(round(float(diferencia_observada), 1), round(float(diferencias_revueltas.std(ddof=1)), 1),
 np.percentile(diferencias_revueltas, [2.5, 97.5]).round(1), round(float(np.abs(diferencias_revueltas).max()), 1))
```

**Interpretación:** la diferencia observada es de unos 675 g. Con las etiquetas revueltas, las diferencias se reparten alrededor de 0 con una desviación estándar de unos 76 g: el 95% central queda entre −152 y +149 g y la mayor que produjo el azar en 10,000 intentos fue de 314 g. La diferencia observada queda a casi nueve veces esa dispersión y ninguna revuelta la igualó, así que el p-value estimado es menor que 1 en 10,000 (un p-value nunca es exactamente 0, y con 10,000 revueltas no se puede precisar más): si H0 fuera cierta (si el sexo no se asociara con la masa), una diferencia así casi nunca se vería. Eso, y nada más, es lo que significa un p-value pequeño.

> 🤔 **Explícalo con tus palabras.** ¿Por qué revolver las etiquetas reproduce lo que pasaría si H0 fuera cierta? Si hubiera el doble de Adelie, ¿el histograma de las revueltas sería más ancho o más angosto, y qué le pasaría al p-value?

**✅ Una explicación posible:** si el sexo no tuviera relación con la masa, la etiqueta «macho» o «hembra» de cada pingüino sería arbitraria y cualquier reparto de etiquetas sería igual de probable; revolverlas genera justo esos repartos. Con el doble de datos, las medias de cada revuelta variarían menos, el histograma sería más angosto y la misma diferencia observada quedaría todavía más lejos: el p-value sería aún menor.

**Ejemplo: comparaciones múltiples y *p-hacking*.** Se simularán 1,000 "estudios" en los que se comparan 20 variables entre dos grupos **idénticos** (H0 es cierta en todas). ¿En qué fracción de estudios aparece al menos un *p* < 0.05?

* La siguiente celda simulará los 1,000 estudios con 20 pruebas *t* cada uno, de forma vectorizada, con su propio generador de números aleatorios (semilla 401) para que el resultado se repita aunque se vuelva a ejecutar la celda.

```python
generador_estudios = np.random.default_rng(401)
grupo_a = generador_estudios.normal(size=(1000, 20, 30))
grupo_b = generador_estudios.normal(size=(1000, 20, 30))
valores_p = stats.ttest_ind(grupo_a, grupo_b, axis=2).pvalue
float((valores_p < 0.05).any(axis=1).mean())
```

Cerca de 64% de los estudios "encuentra algo" (1 − 0.95²⁰ ≈ 0.64, cuenta que supone pruebas independientes, como en esta simulación; con variables relacionadas entre sí saldría otra cifra) sin que exista ningún efecto. Probar muchas variables, subgrupos o versiones de un análisis y reportar solo lo significativo es *p-hacking*. **Decisiones:** definir las preguntas antes de mirar los resultados, reportar todas las pruebas hechas y, si son muchas, corregir con Holm o Bonferroni (`multipletests`).

> 🤔 **Explícalo con tus palabras.** Cada una de las 20 pruebas se equivoca solo 5% de las veces. ¿Por qué, entonces, casi dos de cada tres estudios «encuentran algo»?

**✅ Una explicación posible:** el 5% es por prueba, no por estudio. Cada prueba es una oportunidad más de falsa alarma: la probabilidad de que **ninguna** de las 20 se equivoque es 0.95²⁰ ≈ 0.36, así que la de que al menos una se equivoque es ≈ 0.64.

* La siguiente celda aplicará la corrección de Holm a las 20 pruebas de cada estudio y calculará en qué fracción de estudios queda algún hallazgo.

```python
algun_hallazgo_holm = np.array([multipletests(p_estudio, alpha=0.05, method="holm")[0].any() for p_estudio in valores_p])
float(algun_hallazgo_holm.mean())
```

Con la corrección, la fracción de estudios con algún falso hallazgo vuelve a quedar alrededor de α = 5%. Ojo: la corrección no elimina los falsos hallazgos; solo mantiene en torno a α la probabilidad de cometer alguno en el conjunto de las 20 pruebas. El costo es menos potencia para cada prueba individual: por eso conviene hacer pocas pruebas, bien elegidas.

**Interpretación:** con Holm, solo ≈ 5.6% de los estudios (56 de 1,000) conserva algún falso hallazgo; la diferencia con 5% es ruido de simulación (con pruebas independientes, la cuenta exacta da 1 − (1 − 0.05/20)²⁰ ≈ 4.9%). En estos estudios, donde H0 es cierta en todas las pruebas, Bonferroni da exactamente la misma cifra (56 de 1,000). La idea de las correcciones es **exigir más** a cada prueba cuando se hacen muchas. Bonferroni, la más sencilla y estricta, divide el 0.05 entre el número de pruebas (con 20 pruebas, cada una necesita *p* < 0.0025). Holm hace lo mismo con el p-value más chico y va relajando la exigencia para los siguientes, así que pierde menos potencia cuando sí hay efectos verdaderos (aquí no los hay, y por eso ambas coinciden).

### 2.4 Potencia: ¿cuántos datos hacen falta?

La potencia depende del tamaño del efecto, de *n* y de α. `TTestIndPower` de *statsmodels* resuelve cualquiera de ellos dados los demás. El tamaño del efecto se expresa con la **d de Cohen** (diferencia de medias en unidades de desviación estándar).

**Concepto: d de Cohen.** Es la diferencia de medias medida en **desviaciones estándar** (combinando las de los dos grupos): una d de 0.5 quiere decir que los grupos están separados media desviación estándar. Como referencia aproximada, 0.2 es un efecto pequeño, 0.5 mediano y 0.8 grande.

**Ejemplo:** si dos grupos difieren 5 puntos en un cuestionario de práctica y los puntajes varían unos 10 puntos dentro de cada grupo, d ≈ 0.5.

Al no tener unidades, permite comparar efectos entre variables distintas y es el insumo de los cálculos de potencia.

* La siguiente celda calculará el *n* por grupo necesario para detectar d = 0.5 con 80% de potencia, y la potencia que se tendría con d = 0.3 y 30 por grupo.

```python
potencia = TTestIndPower()
(round(potencia.solve_power(effect_size=0.5, alpha=0.05, power=0.8), 1),
 round(float(potencia.power(effect_size=0.3, nobs1=30, alpha=0.05)), 3))
```

Se necesitan unos 64 casos por grupo para un efecto mediano; con 30 por grupo y un efecto pequeño (d = 0.3), solo se detectaría 1 de cada 5 veces. **Consecuencia:** en un estudio pequeño, "no significativo" casi no informa nada.

> 🤔 **Explícalo con tus palabras.** ¿Por qué, con 30 casos por grupo y d = 0.3, un resultado «no significativo» casi no informa?

**✅ Una explicación posible:** con esa potencia (alrededor de 0.2), aun si el efecto existiera, unos 4 de cada 5 estudios así darían «no significativo». Ese resultado es lo esperable tanto si hay efecto como si no lo hay, así que no ayuda a distinguir entre los dos casos.

📖 **Para profundizar (opcional).** *Las fórmulas de la sección 2.*
* Estadístico *t* para dos medias: $t = \dfrac{\bar{x}_A - \bar{x}_B}{EE(\bar{x}_A - \bar{x}_B)}$, y p-value bilateral: $p = P(|T| \geq |t_{obs}| \mid H_0)$. Se rechaza H0 si $p < \alpha$.
* Con $m$ pruebas, Bonferroni compara cada p-value con $\alpha / m$; Holm ordena los p-values de menor a mayor y compara el *k*-ésimo con $\alpha / (m - k + 1)$.
* d de Cohen con la desviación estándar combinada $s_p$: $$d = \frac{\bar{x}_A - \bar{x}_B}{s_p}, \qquad s_p = \sqrt{\frac{(n_A - 1)s_A^2 + (n_B - 1)s_B^2}{n_A + n_B - 2}}$$

**💡 Ideas clave de la sección 2:**
* Asignar al azar permite hablar de causa; muestrear al azar permite generalizar. Sin ninguno de los dos, se describe una asociación en los datos estudiados.
* Cada prueba tiene una sola H0 y una sola H1, escritas antes de ver los datos; varias preguntas son varias pruebas.
* El p-value mide qué tan sorprendentes serían los datos **si H0 fuera cierta**; no es la probabilidad de H0.
* Con muchas pruebas aparecen falsos hallazgos: se planean pocas pruebas y se corrige (Holm).
* La potencia se planea antes: con *n* pequeño, un "no significativo" casi no informa.

**🎯 Objetivo:** hacer de principio a fin una prueba por permutación, como la de la sección 2, con un caso nuevo. **Qué se espera aprender:** a escribir H0 y H1 antes de mirar el resultado, a construir con código el mundo donde H0 es cierta y a leer un p-value grande sin concluir que los grupos son iguales.

3. **Estadístico de prueba.** Calcula la diferencia observada de masa media: Dream menos Torgersen.

5. **p-value.** Calcula la fracción de revueltas cuya diferencia, en valor absoluto, es al menos tan grande como la observada (la prueba es bilateral). Si quieres, dibuja el histograma de las revueltas con una línea en la diferencia observada.

* La siguiente celda muestra una solución posible.

```python
# Paso 2: los Adelie de Dream y de Torgersen, con su masa.
adelie_islas = pinguinos[(pinguinos["species"] == "Adelie")
                         & pinguinos["island"].isin(["Dream", "Torgersen"])].dropna(subset=["body_mass_g"])
masa_islas = adelie_islas["body_mass_g"].to_numpy()
es_dream = (adelie_islas["island"] == "Dream").to_numpy()

# Paso 3: el estadístico de prueba, Dream menos Torgersen.
diferencia_islas = masa_islas[es_dream].mean() - masa_islas[~es_dream].mean()

# Paso 4: el mundo de H0, con la etiqueta de isla revuelta 10,000 veces.
generador = np.random.default_rng(401)
revueltas_islas = np.empty(10_000)
for i in range(10_000):
    revuelto = generador.permutation(es_dream)
    revueltas_islas[i] = masa_islas[revuelto].mean() - masa_islas[~revuelto].mean()

# Paso 5: p-value bilateral y gráfica de las revueltas.
p_value_islas = float((np.abs(revueltas_islas) >= abs(diferencia_islas)).mean())
ax = sns.histplot(revueltas_islas, bins=50)
ax.axvline(diferencia_islas, color="black")
ax.set(xlabel="diferencia de masa media (g) con las etiquetas de isla revueltas",
       title="Adelie de Dream contra Torgersen (línea negra: diferencia observada)")
plt.show()
{"n Dream": int(es_dream.sum()), "n Torgersen": int((~es_dream).sum()),
 "medias (g)": (round(float(masa_islas[es_dream].mean()), 1), round(float(masa_islas[~es_dream].mean()), 1)),
 "diferencia (g)": round(float(diferencia_islas), 1),
 "95% central de las revueltas (g)": np.percentile(revueltas_islas, [2.5, 97.5]).round(1),
 "p-value": p_value_islas}
```

**H0:** en la población de Adelie de la que vienen estos datos, la masa media de los de Dream es igual a la de los de Torgersen (diferencia 0). **H1:** las dos masas medias son distintas, en cualquier sentido (prueba bilateral).

Al revolver 10,000 veces la etiqueta de isla, el 95% central de las diferencias quedó entre −174 y 169 g, y 8,372 de las 10,000 revueltas dieron una diferencia de 18 g o más, hacia cualquier lado: el p-value es ≈ 0.84.
